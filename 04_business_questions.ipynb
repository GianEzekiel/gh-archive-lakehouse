##  What is the most active repos/branches by push activity

```python
%sql
SELECT
    fact_table.repo_id AS repo_id,
    fact_table.branch_name AS branch_name,
    dim_table.repo_name AS repo_name,
    COUNT(*) AS push_count
FROM main.gh_archive.fact_push_activity AS fact_table
LEFT JOIN main.gh_archive.dim_repo AS dim_table
    ON fact_table.repo_id = dim_table.repo_id
GROUP BY  fact_table.repo_id, fact_table.branch_name, dim_table.repo_name
ORDER BY push_count DESC
LIMIT 10;
```

## What is the longest average PR cycle time, by repo

```python
%sql
SELECT 
    fact_table.repo_id AS repo_id,
    dim_table.repo_name AS repo_name,
    AVG(fact_table.cycle_time_hours) AS average_cycle_time_hours
FROM main.gh_archive.fact_pr_lifecycle AS fact_table
LEFT JOIN main.gh_archive.dim_repo AS dim_table
    ON fact_table.repo_id = dim_table.repo_id
WHERE fact_table.closed_at IS NOT NULL
GROUP BY fact_table.repo_id, dim_table.repo_name
ORDER BY average_cycle_time_hours DESC
LIMIT 10;
```

## Fastest/slowest PR cycle time, by contributor

```python
%sql
SELECT
    dim_actor.actor_id AS actor_id,
    dim_actor.display_login AS display_login,
    AVG(fact_pr_lifecycle.cycle_time_hours) AS average_pr_hours
FROM main.gh_archive.fact_pr_lifecycle AS fact_pr_lifecycle
LEFT JOIN main.gh_archive.dim_actor AS dim_actor
    ON fact_pr_lifecycle.actor_id = dim_actor.actor_id
GROUP BY dim_actor.actor_id, dim_actor.display_login
ORDER BY average_pr_hours DESC
LIMIT 10;
```
